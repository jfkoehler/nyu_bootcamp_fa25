### Webscraping and Beautiful Soup

To begin, we will use `BeautifulSoup` to parse responses containing HTML.  Below, some basic HTML tags are shown -- this is how our responses will be formatted

```python
url = 'https://www.strandbooks.com/books-and-media/childrens/non-fiction.html'
```

```python
r = requests.get(url)
```

```python
r
```

```python
from bs4 import BeautifulSoup
```

```python
soup = BeautifulSoup(r.text)
```

```python
r.text[:1000]
```

```python
soup.find('div', {'class': 'product-info'}).a.text.strip()
```

```python
titles = soup.find_all('div', {'class': 'product-info'})
```

```python
[t.a.text.strip() for t in titles]
```
